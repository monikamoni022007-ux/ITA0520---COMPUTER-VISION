```python

from google.colab import files

uploaded = files.upload()
```

```python
import cv2
import matplotlib.pyplot as plt
```

```python
image_path = list(uploaded.keys())[0]

image = cv2.imread(image_path)

print("Image uploaded successfully:", image_path)
```

```python
blur_image = cv2.GaussianBlur(image, (15, 15), 0)

#image → original image
#	(15, 15) → blur kernel size
#	0 → OpenCV automatically calculates the Gaussian standard deviation */
```

```python
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")
plt.show()

```

```python
plt.imshow(cv2.cvtColor(blur_image, cv2.COLOR_BGR2RGB))
plt.title("Gaussian Blurred Image")
plt.axis("off")
plt.show()
```

```python
plt.figure(figsize=(10, 4))

# Original Image
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

# Gaussian Blurred Image
plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(blur_image, cv2.COLOR_BGR2RGB))
plt.title("Gaussian Blurred Image")
plt.axis("off")

plt.show()


```
